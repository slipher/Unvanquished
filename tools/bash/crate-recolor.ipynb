```python
A = 0.055
phi = 12.92
X = 0.04045
G = 2.4

dat = imageio.imread('C:/unv/Unvanquished/pkg/bash-assets_src.dpkdir/models/vil_p_crate06_d.png')
dat = dat.astype(float) / 255.
def srgb2linear(color):
    linearMask = color < X
    linear = color * (1 / phi)
    gamma = ((color + A) / (1 + A)) ** G
    return linearMask * linear + (1 - linearMask) * gamma
def linear2srgb(color):
    linearMask = color < X / phi
    linear = color * phi
    gammaInv = color ** (1/G) * (1 + A) - A
    return linearMask * linear + (1 - linearMask) * gammaInv

%matplotlib inline
from matplotlib import pyplot as plt
roundtrip = linear2srgb(srgb2linear(dat))
plt.imshow(roundtrip)
```

```python
import numpy as np

srgb = linear2srgb(dat)
gray = np.inner(srgb, [0.30, 0.59, 0.11])
plt.imshow(gray)
#already succeeded at making everything green???
# but did the srgb conversion backwards
```
